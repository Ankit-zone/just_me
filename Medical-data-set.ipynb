```python
import pandas as pd
```

```python
df=pd.read_csv(r'D:\Users\ankit\medical.csv')
df
```

```python
x=df['Blood_Pressure'].mean()
df.fillna({'Blood_Pressure':x},inplace=True)
df
```

```python
x=df['Age'].mean()
df.fillna({'Age':x},inplace=True)
df
```

```python
x=df['Cholesterol'].mean()
df.fillna({'Cholesterol':x},inplace=True)
df
```

```python
df.replace('N', 'No', inplace=True)
df.replace('Y', 'Yes', inplace=True)
df
```

```python
# Replaces the literal string 'NaN' with Python's None object in the 'Smoker' column
df['Gender'] = df['Gender'].replace('MALE', 'Male')
df['Gender'] = df['Gender'].replace('FEMALE', 'Female')
df
```

```python
df.drop_duplicates(inplace = True)
df
```

```python
df.to_excel('medical.', index=False)
```
