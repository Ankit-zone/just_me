```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
x=np.linspace(0,5,11)
y=x**2
```

```python
x
```

```python
y
```

```python
plt.plot(x,y,"r")
plt.xlabel('x axis here')
plt.ylabel('y axis here')
plt.title('String Title here')
plt.show()
```

```python
plt.subplot(1,2,1)
plt.plot(x,y,'r--')
plt.subplot(1,2,2)
plt.plot(y,x,'g*-');
```

```python
fig=plt.figure()
axes=fig.add_axes([0.1,0.1,0.8,0.8])
axes.plot(x,y,'b')
axes.set_xlabel('set x label')
axes.set_ylabel('set y label')
axes.set_title('Set Title')
```

```python
fig =plt.figure()
axes1=fig.add_axes([0.1,0.1,0.8,0.8])
axes2=fig.add_axes([0.2,0.5,0.4,0.3])
axes1.plot(x,y,'b')
axes1.set_xlabel(' x_label_axes2')
axes1.set_ylabel(' y_label_axes2')
axes1.set_title('axes1 Title')
axes2.plot(x,y,'r')
axes2.set_xlabel(' X_label_axes2')
axes2.set_ylabel(' Y_label_axes2')
axes2.set_title('axes2  Title')
```

```python
fig,axes=plt.subplots()
axes.plot(x,y,'r')
axes.set_xlabel('x')
axes.set_ylabel('y')
axes.set_title('title')
```

```python
fig,axes=plt.subplots(nrows=1,ncols=2)
```

```python
for ax in axes:
    ax.plot(x,y,'b')
    ax.set_xlabel('x')
    ax.set_ylabel('y')
    ax.set_title('title')
fig
```

```python
fig,axes=plt.subplots(nrows=1,ncols=2)
for ax in axes:
    ax.plot(x,y,'g')
    ax.set_xlabel('x')
    ax.set_ylabel('y')
    ax.set_title('title')
fig
plt.tight_layout()
```

```python
fig,ax=plt.subplots(figsize=(12,3))

ax.plot(x,y,'g')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_title('title');
```

```python
fig=plt.figure()
ax=fig.add_axes([0,0,1,1])
ax.plot(x,x**2,label='x**2')
ax.plot(x,x**3,label='x**2')
ax.legend()
```

```python
fig=plt.figure()
ax=fig.add_axes([0,0,1,1])
ax.plot(x,x**2,label='x**2')
ax.plot(x,x**3,label='x**2')
ax.legend(loc=1)
ax.legend(loc=2)
ax.legend(loc=3)
ax.legend(loc=4)
```
