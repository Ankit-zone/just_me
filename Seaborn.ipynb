```python
import seaborn as sns
```

```python
import matplotlib.pyplot as plt
```

```python
sns.set_theme()
sns.get_dataset_names()
```

```python
tips = sns.load_dataset("tips")
print(tips.head())
```

```python
sns.relplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue = "smoker",
    size = "size",
    style = "smoker"
)
```

```python
sns.relplot(
    data=tips,
    x="total_bill",
    y="tip",
    kind = "line",
)
```

```python
# Scatter plot
tips = sns.load_dataset("tips")
sns.scatterplot(
    data = tips,
    x = "total_bill",
    y = "tip",
    hue = "time"
)
```

```python
#Line plot

tips = sns.load_dataset("tips")
sns.lineplot(
    data = tips,
    x = "total_bill",
    y = "tip"
)
```

```python
flights = sns.load_dataset("flights")

print(flights.head())
```

```python
sns.lineplot(
    data = flights,
    x = "year",
    y = "passengers"
)
```

```python
print(tips.head())
```

```python
#Bar plot

sns.barplot(
    data = tips,
    x = "day",
    y = "tip",
    hue = "sex"
)
```

```python
#Box plot 

sns.boxplot(
    data = tips,
    x = "day",
    y = "tip"
)
```

```python
penguins = sns.load_dataset("penguins")
```

```python
print(penguins.head())
```

```python
sns.histplot(
    data = penguins,
    x = "body_mass_g",
    bins = 25
)
```

```python
flights_heat = flights.pivot(index = "month", columns = "year", values = "passengers")
print(flights_heat)

sns.heatmap(
    flights_heat,
    cmap = "coolwarm",
    annot = True,
    fmt = "d"
)
plt.title("HeatMap for passengers")
plt.tight_layout()
```

```python
fig , ax = plt.subplots()

sns.lineplot(
    data = tips,
    x = "day",
    y = "total_bill",
    hue = "sex",
    marker = "o",
    ax=ax,
    errorbar = None
)
ax.set_title("Avg Bill value for each days")
ax.set_xlabel("Day")
ax.set_ylabel("Total Bill")

fig.tight_layout()
```
